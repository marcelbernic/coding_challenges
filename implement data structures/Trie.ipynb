```python
class TrieNode:
    def __init__(self, char = ""):
        self.char = char
        self.is_word = False
        self.children = {}
```

```python
class Trie:

    def __init__(self):
        pass
        

    def insert(self, word: str) -> None:
        pass
        

    def search(self, word: str) -> bool:
        pass

    def startsWith(self, prefix: str) -> bool:
        pass
```

```python
# Solution
class Trie:

    def __init__(self):
        self.root = TrieNode()
        

    def insert(self, word: str) -> None:
        current = self.root
        for char in word:
            if char not in current.children:
                current.children[char] = TrieNode(char)
                
            current = current.children[char]
        current.is_word = True
        

    def search(self, word: str) -> bool:
        current = self.root
        for char in word:
            if char not in current.children:
                return False
            current = current.children[char]
        
        return current.is_word
        

    def startsWith(self, prefix: str) -> bool:
        current = self.root
        for char in prefix:
            if char not in current.children:
                return False
            current = current.children[char]
        
        return True
        
```
