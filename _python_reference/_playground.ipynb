```python
from collections import Counter

def count_and_sort_elements(lst):
    id_counts = Counter(lst)
    sorted_ids = sorted(id_counts.items(), key=lambda x: x[1], reverse=False)

    return sorted_ids


print(count_and_sort_elements(['a', 'x', 'q', 'z', 't', 'o', 'o']))
```

```python
import time

# Create a TTL cache with a max size of 100 and a time-to-live of 300 seconds
cache = TTLCache(maxsize=4096, ttl=15)

@cached(cache)
def expensive_function(x):
    time.sleep(1)
    print(f'Calling the method for -> {x}')
    # print(f'Not cached for {x}') if not cache.get(x) else print(f'Cached for {x}')
    return x * x

# Warm up the cache by calling the function with some arguments
# a = expensive_function(1)
# b = expensive_function(2)

# Measure the execution time for the first call
start_time = time.time()
res = expensive_function(1)
end_time = time.time()
print(f"RES:{res} => {end_time - start_time:.6f} seconds")

# print('Sleeping for 1 second')
# time.sleep(1) # in theory the cache is invalidated

# Measure the execution time for the second call
start_time = time.time()
res = expensive_function(1)
end_time = time.time()
print(f"RES:{res} => {end_time - start_time:.6f} seconds")

# Measure the execution time for the second call
start_time = time.time()
res = expensive_function(2)
end_time = time.time()
print(f"RES:{res} => {end_time - start_time:.6f} seconds")

# Measure the execution time for the first call with a new argument
start_time = time.time()
res = expensive_function(1)
end_time = time.time()
print(f"RES:{res} => {end_time - start_time:.6f} seconds")

# Measure the execution time for the first call with a new argument
start_time = time.time()
res = expensive_function(1)
end_time = time.time()
print(f"RES:{res} => {end_time - start_time:.6f} seconds")

start_time = time.time()
res = expensive_function(3)
end_time = time.time()
print(f"RES:{res} => {end_time - start_time:.6f} seconds")

# Measure the execution time for the second call
start_time = time.time()
res = expensive_function(2)
end_time = time.time()
print(f"RES:{res} => {end_time - start_time:.6f} seconds")

start_time = time.time()
res = expensive_function(3)
end_time = time.time()
print(f"RES:{res} => {end_time - start_time:.6f} seconds")

print('Sleeping for 15 seconds')
time.sleep(15) # in theory the cache is invalidated

# Measure the execution time for the first call
start_time = time.time()
res = expensive_function(1)
end_time = time.time()
print(f"RES:{res} => {end_time - start_time:.6f} seconds")

# Measure the execution time for the first call
start_time = time.time()
res = expensive_function(2)
end_time = time.time()
print(f"RES:{res} => {end_time - start_time:.6f} seconds")

# Measure the execution time for the first call with a new argument
start_time = time.time()
res = expensive_function(3)
end_time = time.time()
print(f"RES:{res} => {end_time - start_time:.6f} seconds")
```
